```python
import json
import time
import random
from datetime import datetime
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers='localhost:9092',
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

cities = ["New York", "San Francisco", "Chicago", "Boston", "Seattle"]

def generate_ride_event():
    return {
        "ride_id": random.randint(1000000, 9999999),
        "city": random.choice(cities),
        "timestamp": datetime.now().isoformat(),
        "distance_km": round(random.uniform(1.0, 15.0), 2),
        "fare_usd": round(random.uniform(5.0, 40.0), 2)
    }

print("Producing ride events...")

while True:
    event = generate_ride_event()
    producer.send("uber-rides", event)
    print("Sent:", event)
    time.sleep(1)
```
