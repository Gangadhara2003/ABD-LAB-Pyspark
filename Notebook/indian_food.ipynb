1. Start Spark and Load Dataset

```python
import pyspark
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
filepath = r"D:\ABD-Lab (BDA)\data\indian_food.csv"
```

```python
spark = SparkSession.builder \
    .appName("Indian Food Analysis") \
    .master("local[*]") \
    .getOrCreate()

spark
```

```python
df = spark.read.csv(
    "D:\ABD-Lab (BDA)\data\indian_food.csv",
    header=True,
    inferSchema=True
)

df.show(10)
```

```python
df.printSchema()
```

2. Find How Many Unique Dishes Are Present

```python
unique_dishes = df.select("name").distinct().count()

print("Number of unique dishes:", unique_dishes)
```

3. Which State Has More Dishes?

```python
state_dish_count = df.groupBy("state").count().orderBy(desc("count"))
state_dish_count.show()
```

```python
state_dish_count.limit(1).show()
```

4. How Many Dishes Are From Karnataka?

```python
karnataka_count = df.filter(col("state") == "Karnataka").count()

print("Number of dishes from Karnataka:", karnataka_count)
```
